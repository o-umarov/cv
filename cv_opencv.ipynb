```python
import cv2
```

```python
from google.colab.patches import cv2_imshow

image = cv2.imread('/content/catimage.jpg')
cv2_imshow(image)
```

```python
print(image.shape) # (350 - height[0], 700 - width,[1] 3 - color channels[2])
print(f"Image Height: {image.shape[0]}")
print(f"Image Width: {image.shape[1]}")
print(f"Image Color Channels: {image.shape[2]}")
# Color Scheme: BGR - Blue Green Red
```

```python
(B, G, R) = image[90, 60]
print(f"Blue: {B}, Green: {G}, Red: {R}")
```

```python
gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
cv2_imshow(gray_image)
```

```python
resized = cv2.resize(gray_image, (400, 400))
cv2_imshow(resized)
```

```python
cropped = gray_image[110:300, 250:550]
cv2_imshow(cropped)
```

```python
rotated = cv2.rotate(resized, cv2.ROTATE_90_COUNTERCLOCKWISE)
cv2_imshow(rotated)
```

```python
flipped = cv2.flip(resized, 1)
cv2_imshow(flipped)
```

```python
cv2.imwrite("resized.jpg", cropped )
```
